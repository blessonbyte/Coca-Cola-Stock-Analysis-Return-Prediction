## Task 7: Compare two groups

**Question.** Do late orders get worse reviews?

**Plan.** One cell that produces: `review_late`, `review_on_time`.

**A weak prompt:** *"Prove that late delivery causes bad reviews."* It asks the AI to confirm your belief. The data show an association, not a cause.

**The prompt, written for you.** Copy it into Gemini or Claude.

```text
Context: DataFrame `orders` has late ('late' or 'on time') and review_score (1 to 5).
Task: Compare late and on-time orders: the number of orders, the average review score, and the share of bad reviews (score 2 or below).
Constraints: Use groupby. Round to 2 decimals.
Format: One cell: one small table, one row per group.
Check: Store the average scores in `review_late` and `review_on_time`.
```

**Paste the AI's code below.** Before you run it, check that it:
- Groups by late
- Shows the size of each group
- Bad review is 2 or below, as asked

````python
# Paste the AI's code here. Read it before you run it.
# Handle the column name (using 'late_status' from our previous step, or 'late' as a fallback)
late_col = 'late_status' if 'late_status' in orders.columns else 'late'

# Create a 1/0 column for bad reviews (1 if score is 2 or below, 0 otherwise)
# Taking the mean of a 1/0 column automatically calculates the share (percentage)
orders['bad_review'] = np.where(orders['review_score'] <= 2, 1, 0)

# Group by the delivery status and calculate our three required metrics
comparison_table = orders.groupby(late_col).agg(
    n_orders=('review_score', 'count'),
    avg_review_score=('review_score', 'mean'),
    share_bad_reviews=('bad_review', 'mean')
).round(2) # Round all results in the table to 2 decimals

# Display the final comparison table
display(comparison_table)

# Store the average scores for the check constraint using .loc[row, column]
review_late = comparison_table.loc['late', 'avg_review_score']
review_on_time = comparison_table.loc['on time', 'avg_review_score']

# Simple check to confirm the variables are stored correctly
print(f"\nCheck -> Late Avg: {review_late}, On-Time Avg: {review_on_time}")
````

## Task 7b: Do ANOVA, analysis of variance

**Q: Is it worth investing in faster or more reliable delivery to improve our ratings?**

Context: DataFrame `orders` has late ('late' or 'on time') and review_score (1 to 5, some missing).
Task: Run a one-way ANOVA to test whether the average review score differs between late and on-time orders.
Constraints: Use scipy.stats.f_oneway. Drop missing review scores. Do not change orders.
Format: One cell. Print the F statistic, the p-value and whether it is significant at 5%, each with a label.
Check: Store the results in `f_stat` and `p_value`.

**Explain.** Write one sentence for a manager, with both averages and the size of each group.

````python
# Paste the AI's code here. Read it before you run it.

import scipy.stats as stats

# Handle the column name based on previous steps (fallback to 'late')
late_col = 'late_status' if 'late_status' in orders.columns else 'late'

# Create two separate, temporary lists of review scores, dropping missing values (NaN)
# 1. Scores for orders that were on time
scores_on_time = orders[orders[late_col] == 'on time']['review_score'].dropna()
# 2. Scores for orders that were late
scores_late = orders[orders[late_col] == 'late']['review_score'].dropna()

# Run the one-way ANOVA test comparing the two groups
# This calculates the F-statistic and the p-value
f_stat, p_value = stats.f_oneway(scores_on_time, scores_late)

# Determine if the result is statistically significant at the 5% level (p < 0.05)
is_significant = "Yes (Reject Null)" if p_value < 0.05 else "No (Fail to Reject)"

# Print the results clearly
print(f"F-Statistic: {f_stat:.2f}")
print(f"P-Value: {p_value:.5f}")
print(f"Significant at 5%?: {is_significant}")

# Simple check to confirm the variables are stored
print(f"\nCheck -> Stored f_stat: {f_stat:.2f}, p_value: {p_value:.5f}")
````

````python
# Paste the AI's code here. Read it before you run it.
import matplotlib.pyplot as plt

# 1. Group by state and calculate both the average delivery days and total order count
state_stats = orders.groupby('customer_state')['delivery_days'].agg(['mean', 'count'])

# 2. Filter the table to only keep states that have 20 or more orders
valid_states = state_stats[state_stats['count'] >= 20]

# 3. Sort by the average delivery days (ascending=True pushes the longest times to the top of the chart)
sorted_states = valid_states.sort_values(by='mean', ascending=True)

# 4. Identify the slowest state (it will be the very last item in our ascending list)
slowest_state = sorted_states.index[-1]

# 5. Create the horizontal bar chart
plt.figure(figsize=(10, 6))

# plt.barh creates horizontal bars. Y-axis is the state names (index), X-axis is the average days ('mean')
plt.barh(sorted_states.index, sorted_states['mean'], color='orange', edgecolor='black')

# Add a clear title stating what the chart shows
plt.title('Average Delivery Time by State (Min. 20 Orders)')
# Label the x-axis with the unit of measurement
plt.xlabel('Average Delivery Days')
# Label the y-axis
plt.ylabel('Customer State')

# Render the plot
plt.show()

# Print the required check variable
print(f"Slowest state (>= 20 orders): {slowest_state}")

# Simple check to confirm the logic works by checking if slowest_state matches the max mean
max_days = sorted_states['mean'].max()
print(f"Check -> The average for {slowest_state} is {max_days:.1f} days.")
````

**Question.** Does what customers paid match price plus freight?

**Plan.** One cell that produces: `naive_rows`, `n_mismatch`.

**A weak prompt:** *"Join payments to orders and total the revenue."* A direct join repeats an order once for each payment row, so revenue is overstated.

**Write your prompt.** Use the five headings. Hints:

- **Context:** orders: one row per order; payments: one or more rows per order, with payment_value.
- **Task:** First, show what a direct join does to the row count. Then total payments per order, join, and compare with price + freight.
- **Constraints:** Aggregate before joining; validate one-to-one; a difference of 0.01 or more counts.
- **Format:** One cell; print the counts with labels.
- **Check:** Store the direct join's rows in `naive_rows` and the orders that differ in `n_mismatch`.

````python
prompt = """
Context:
Task:
Constraints:
Format:
Check:
"""
check_prompt(prompt, table="orders", must=['naive_rows', 'n_mismatch'])
````

**Paste the AI's code below.** Before you run it, check that it:
- Sums payments per order before joining
- validate='1:1' on the second join
- Rounds the difference before comparing

````python
# Paste the AI's code here. Read it before you run it.
import pandas as pd
import numpy as np

# 1. THE TRAP: Demonstrate a direct (naive) join without aggregating first
naive_join = orders.merge(payments, on='order_id', how='left')
# Store the inflated row count from the naive join
naive_rows = naive_join.shape[0]

# 2. THE SOLUTION: Aggregate the payments table so there is only one row per order
# We group by order_id and sum the payment values, then reset the index to keep it as a DataFrame
payments_agg = payments.groupby('order_id')['payment_value'].sum().reset_index()

# 3. Join the aggregated payments safely back to the orders table
# validate='1:1' forces pandas to crash if either side still has duplicates
orders_checked = orders.merge(payments_agg, on='order_id', how='left', validate='1:1')

# 4. Compare what was paid versus what was owed (Price + Freight)
# Fill missing payments with 0 just in case an order had no payment record
total_paid = orders_checked['payment_value'].fillna(0)
total_owed = orders_checked['price'] + orders_checked['freight']

# Calculate the absolute difference (handles both overpayments and underpayments)
payment_diff = (total_paid - total_owed).abs()

# Count how many orders have a mismatch of 1 cent or more
n_mismatch = (payment_diff >= 0.01).sum()

# Print the results with clear labels
print(f"Original orders count: {orders.shape[0]}")
print(f"Naive join row count (Inflated!): {naive_rows}")
print(f"Orders with payment mismatch (>= 0.01): {n_mismatch}")

# Simple check to confirm the variables were stored and logic held up
print(f"\nCheck -> Naive rows: {naive_rows}, Mismatches: {n_mismatch}")
print(f"Did aggregation prevent row explosion? {orders_checked.shape[0] == orders.shape[0]}")
````
